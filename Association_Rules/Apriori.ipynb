```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("../Datasets/Market_Basket_Optimisation.csv",header=None)
```

```python
dataset.head()
```

```python
len(dataset)
```

```python
transactions=[]

for i in range(0,7501):
    transaction=[]
    for j in range(0,20):
        item=dataset.values[i,j]

        if pd.notna(item):
            transaction.append(str(item))
    transactions.append(transaction)
```

```python
from apyori import apriori

rules=apriori(transactions=transaction,min_support=0.003,min_confidence=0.2,min_lift=3,min_length=2,max_length=2)
print(rules)
```
