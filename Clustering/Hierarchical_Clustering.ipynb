```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
dataset=pd.read_csv("../Datasets/customer_clustering_dataset.csv")
dataset
```

```python
X=dataset.iloc[:,[3,4]].values
```

```python
import scipy.cluster.hierarchy as sch
dendogram=sch.dendrogram(sch.linkage(X,method="ward"))
plt.title('Dendogram')
plt.xlabel("Customers")
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

hc=AgglomerativeClustering(n_clusters=3,metric="euclidean",linkage='ward')
y_means=hc.fit_predict(X)
```

# **Visualize the clusters**

```python
plt.scatter(X[y_means==0,0],X[y_means==0,1],c="red",label='Cluster 1')
plt.scatter(X[y_means==1,0],X[y_means==1,1],c='green',label='Cluster 2')
plt.scatter(X[y_means==2,0],X[y_means==2,1],c='blue',label='Cluster 3')
plt.title("Customer Segmentation")
plt.xlabel("Annual Income ($)")
plt.ylabel("Spending Score(1-100)")
plt.legend()
plt.show()
```
